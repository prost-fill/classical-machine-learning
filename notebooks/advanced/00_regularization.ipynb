## 0. Imports

````python
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import numpy as np
import joblib
import warnings
warnings.filterwarnings('ignore')
dayofweek = '../../datasets/dayofweek.csv'
````

## 1. Preprocessing

1. Read the file `dayofweek.csv` that you used in the previous day to a dataframe.
2. Using `train_test_split` with parameters `test_size=0.2`, `random_state=21` get `X_train`, `y_train`, `X_test`, `y_test`. Use the additional parameter `stratify`.

````python
df = pd.read_csv(dayofweek, index_col=0)
df
````

````python
X = df.drop('dayofweek', axis=1)
y = df['dayofweek']
````

````python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=21, stratify=y
)
````

## 2. Logreg regularization

### a. Default regularization

1. Train a baseline model with the only parameters `random_state=21`, `fit_intercept=False`.
2. Use stratified K-fold cross-validation with `10` splits to evaluate the accuracy of the model


The result of the code where you trained and evaluated the baseline model should be exactly like this (use `%%time` to get the info about how long it took to run the cell):

```

```

````python
def func (model, X = X_train, y = y_train, n_folds = 10):
    CrossVal = StratifiedKFold(n_splits=n_folds, shuffle=True, random_state=21)
    scores = []
    for train, val in CrossVal.split(X,y):
        X_train = X.iloc[train]
        y_train = y.iloc[train]
        X_v = X.iloc[val]
        y_v = y.iloc[val]
        model.fit(X_train,y_train )
        y_pred = model.predict(X_v)
        train_pred = model.predict(X_train)
        train_score = accuracy_score(y_train, train_pred)
        val_score = accuracy_score(y_v, y_pred)
        scores.append(val_score)
        print(f"train - {train_score:.5f}   |   valid - {val_score:.5f}")
    print(f"Average accuracy on crossval is {np.mean(scores):.5f}")
    print(f"Std is {np.std(scores):.5f}")
````

````python
%%time
lr = LogisticRegression(random_state=21, fit_intercept=False)
func(model = lr)
````

````python
%%time
lr_none = LogisticRegression(random_state=21, fit_intercept=False, penalty=None, max_iter=1000)
func(model=lr_none)
````

## 6. Predictions

1. Choose the best model and use it to make predictions for the test dataset.
2. Calculate the final accuracy.
3. Analyze: for which weekday your model makes the most errors (in % of the total number of samples of that class in your test dataset).
4. Save the model.

````python
model = RandomForestClassifier(n_estimators=100, max_depth=20, random_state=21)
func(model = model)
````

````python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
accuracy
````

````python
result = pd.DataFrame({"true": y_test, "prediction": y_pred })


for el in result["true"].unique():
    data = result[result["true"] == el]
    total = len(data)
    errors = (data["true"] != data["prediction"]).sum()
    errors_rate = errors/total*100
    print(f"{el}:{errors_rate:.2f}%")
````

````python
joblib.dump(model, "model.plk")
````
